## Feature Engineering

```python
data['Country'].value_counts()
```

```python
# We have 111 country so we take top 10 country
top_country = data['Country'].value_counts().index[0:11].to_list()
```

```python
# We convert 111 countries into top 10 countries + other 
def group_countries(country):
    if country in top_country:
        return country
    else:
        return 'Other'
```

```python
data['Group_Country']= data['Country'].apply(group_countries)
```

```python
data['Group_Country'].value_counts()
```

## Train Test Split & Create Pipeline

```python
from sklearn.model_selection import train_test_split

skewed_col = ['Study_Hours']
other_num_col = ['Age', 'Avg_Daily_Usage_Hours', 'Daily_Unlocks','Physical_Activity_Hours', 'Sleep_Hours_Per_Night']
ordinal_col = ['Stress_Level']  
ohe_col = ["Gender", "Academic_Level", "Most_Used_Platform", "Purpose_Of_Use", "Group_Country"]

feature_col = skewed_col + other_num_col + ordinal_col + ohe_col
x = data.drop(['Mental_Health_Score','Country'],axis=1)
y = data['Mental_Health_Score']

x_train, x_test,y_train,y_test = train_test_split(x,y, test_size= 0.3)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import FunctionTransformer,StandardScaler,OrdinalEncoder,OneHotEncoder
from sklearn.compose import ColumnTransformer


skew_pipeline = Pipeline(steps=[
    ('log transform',FunctionTransformer(np.log1p)),
    ('scaler', StandardScaler())
])

other_num_pipeline = Pipeline(steps=[
    ('scaler', StandardScaler())
])

ordinal_pipeline = Pipeline(steps=[
    ('ordinal', OrdinalEncoder(categories=[['Low','Medium','High','Very High']]))
])

ohe_pipeline = Pipeline(steps=[
    ('Ohe', OneHotEncoder(handle_unknown='ignore'))
])

Preprocessor = ColumnTransformer(transformers=[
    ('Skewed', skew_pipeline, skewed_col),
    ('Numerical',other_num_pipeline, other_num_col),
    ('Ordinal',ordinal_pipeline, ordinal_col),
    ('One Hot Encoding',ohe_pipeline, ohe_col)
])
```

## Linear Regression

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

lr_pipeline = Pipeline(steps=[
    ('Preprocessing', Preprocessor),
    ('Regression',LinearRegression())
])

lr_pipeline.fit(x_train,y_train)
lr_pre = lr_pipeline.predict(x_test)
lr_pre_train = lr_pipeline.predict(x_train)

lr_r2_testing = r2_score(y_test,lr_pre)
lr_r2_training = r2_score(y_train, lr_pre_train) 
lr_mean_absolute_error = mean_absolute_error(y_test,lr_pre)
lr_mean_squared_error = mean_squared_error(y_test,lr_pre)

print(f"Accuracy of Training: {lr_r2_training}")
print(f"Accuracy of Testing: {lr_r2_testing}")
print(f"Mean_absolute_error: {lr_mean_absolute_error}")
print(f"mean_squared_error: {lr_mean_squared_error}")
```

## Random Forest

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

rf_pipeline = Pipeline(steps=[
    ('Preprocessing', Preprocessor),
    ('Classifier', RandomForestRegressor(random_state=42))
])

rf_pipeline.fit(x_train,y_train)
rf_pre = rf_pipeline.predict(x_test)
rf_pre_train = rf_pipeline.predict(x_train)

rf_r2_testing = r2_score(y_test,rf_pre)
rf_r2_training = r2_score(y_train, rf_pre_train)
rf_mean_absolute_error = mean_absolute_error(y_test,rf_pre)
rf_mean_square_error = mean_squared_error(y_test,rf_pre)

print(f"Accuracy of Training: {rf_r2_training}")
print(f"Accuracy of Testing: {rf_r2_testing}")
print(f"Mean_absolute_error: {rf_mean_absolute_error}")
print(f"mean_squared_error: {rf_mean_absolute_error}")
```

## Hyperparameter Tuning

```python
from sklearn.model_selection import RandomizedSearchCV

parameter = {
    'Classifier__n_estimators': [100,200,300],
    'Classifier__max_depth': [5,10,20],
    'Classifier__min_samples_split': [2,5,10],
    'Classifier__min_samples_leaf': [1,2,3]
}

random_search = RandomizedSearchCV(
    estimator=rf_pipeline,
    param_distributions=parameter,
    n_iter= 15,
    cv= 5,                           #Cross validation
    scoring='r2',
    random_state=42,
    n_jobs= -1                       #Use all free processor
)

random_search.fit(x_train,y_train)
```

```python
random_search.best_params_
```

```python
rf_best_pipeline = random_search.best_estimator_
rf_best_pre = rf_best_pipeline.predict(x_test)
rf_best_pre_train = rf_best_pipeline.predict(x_train)

rf_best_r2_testing = r2_score(y_test,rf_best_pre)
rf_best_r2_training = r2_score(y_train, rf_best_pre_train)
rf_best_mean_absolute_error = mean_absolute_error(y_test,rf_best_pre)
rf_best_mean_square_error = mean_squared_error(y_test,rf_best_pre)

print(f"Accuracy of Training: {rf_best_r2_training}")
print(f"Accuracy of Testing: {rf_best_r2_testing}")
print(f"Mean_absolute_error: {rf_best_mean_absolute_error}")
print(f"mean_squared_error: {rf_best_mean_absolute_error}")
```
